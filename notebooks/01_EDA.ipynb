```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../dataset/cleaned_traffic_data.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print(df.info())
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
print(df["signal_status"].value_counts())

df["signal_status"].value_counts().plot(kind="bar")
plt.title("Signal Status Distribution")
plt.xlabel("Signal Status")
plt.ylabel("Count")
plt.show()
```

```python
print(df["weather_condition"].value_counts())

df["weather_condition"].value_counts().plot(kind="bar")
plt.title("Weather Condition Distribution")
plt.xlabel("Weather Condition")
plt.ylabel("Count")
plt.show()
```

```python
plt.hist(df["traffic_volume"], bins=30)
plt.title("Traffic Volume Distribution")
plt.xlabel("Traffic Volume")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(df["avg_vehicle_speed"], bins=30)
plt.title("Average Vehicle Speed Distribution")
plt.xlabel("Average Vehicle Speed")
plt.ylabel("Frequency")
plt.show()
```

```python
print(df["location_id"].value_counts().sort_index())

df["location_id"].value_counts().sort_index().plot(kind="bar")
plt.title("Traffic Records by Location")
plt.xlabel("Location ID")
plt.ylabel("Count")
plt.show()
```

```python
df.groupby("location_id")["traffic_volume"].mean().plot(kind="bar")
plt.title("Average Traffic Volume by Location")
plt.xlabel("Location ID")
plt.ylabel("Average Traffic Volume")
plt.show()
```

```python
vehicle_columns = [
    "vehicle_count_cars",
    "vehicle_count_trucks",
    "vehicle_count_bikes"
]

df[vehicle_columns].mean().plot(kind="bar")
plt.title("Average Vehicle Count by Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Count")
plt.show()
```

```python
plt.scatter(df["traffic_volume"], df["avg_vehicle_speed"])
plt.title("Traffic Volume vs Average Vehicle Speed")
plt.xlabel("Traffic Volume")
plt.ylabel("Average Vehicle Speed")
plt.show()
```

```python
df.groupby("weather_condition")["traffic_volume"].mean().plot(kind="bar")
plt.title("Average Traffic Volume by Weather Condition")
plt.xlabel("Weather Condition")
plt.ylabel("Average Traffic Volume")
plt.show()
```

```python
print(df["accident_reported"].value_counts())

df["accident_reported"].value_counts().plot(kind="bar")
plt.title("Accident Reported Distribution")
plt.xlabel("Accident Reported")
plt.ylabel("Count")
plt.show()
```

```python
plt.scatter(df["temperature"], df["traffic_volume"])
plt.title("Temperature vs Traffic Volume")
plt.xlabel("Temperature")
plt.ylabel("Traffic Volume")
plt.show()
```

```python
numeric_columns = [
    "location_id",
    "traffic_volume",
    "avg_vehicle_speed",
    "vehicle_count_cars",
    "vehicle_count_trucks",
    "vehicle_count_bikes",
    "temperature",
    "humidity",
    "accident_reported"
]

correlation = df[numeric_columns].corr()

print(correlation)

plt.figure(figsize=(10, 7))
plt.imshow(correlation, aspect="auto")
plt.colorbar()
plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=90)
plt.yticks(range(len(correlation.columns)), correlation.columns)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```
