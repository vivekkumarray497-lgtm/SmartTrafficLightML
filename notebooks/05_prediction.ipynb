```python
import pandas as pd
import joblib

model = joblib.load("../model/random_forest_model.pkl")

print("Random Forest model loaded successfully!")
```

```python
weather_mapping = {
    "Sunny": 0,
    "Cloudy": 1,
    "Rainy": 2,
    "Foggy": 3,
    "Windy": 4
}

signal_mapping = {
    0: "Red",
    1: "Yellow",
    2: "Green"
}

print("Mappings created successfully!")
```

```python
input_data = pd.DataFrame([{
    "location_id": 4,
    "traffic_volume": 600,
    "avg_vehicle_speed": 40,
    "vehicle_count_cars": 500,
    "vehicle_count_trucks": 30,
    "vehicle_count_bikes": 70,
    "weather_condition": weather_mapping["Sunny"],
    "temperature": 25,
    "humidity": 60,
    "accident_reported": 0,
    "hour": 8,
    "minute": 30,
    "day": 1,
    "month": 1
}])

print("Input data:")
display(input_data)
```

```python
prediction = model.predict(input_data)

predicted_signal = signal_mapping[prediction[0]]

print("Predicted Signal:", predicted_signal)
```

```python
print("=" * 40)
print("SMART TRAFFIC LIGHT PREDICTION")
print("=" * 40)

print("Location ID:", input_data["location_id"].iloc[0])
print("Traffic Volume:", input_data["traffic_volume"].iloc[0])
print("Vehicle Speed:", input_data["avg_vehicle_speed"].iloc[0])
print("Temperature:", input_data["temperature"].iloc[0])
print("Humidity:", input_data["humidity"].iloc[0])
print("Accident Reported:", input_data["accident_reported"].iloc[0])
print("Predicted Signal:", predicted_signal)
```
